### Acquire image with epics

```python
nx = epics.caget('CAX:B:BASLER01:image1:ArraySize1_RBV')
ny = epics.caget('CAX:B:BASLER01:image1:ArraySize0_RBV')
data = epics.caget('CAX:B:BASLER01:image1:ArrayData')
image = np.array(data).reshape((nx, ny))
```
